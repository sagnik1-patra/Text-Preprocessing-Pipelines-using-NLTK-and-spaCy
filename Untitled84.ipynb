```python
import os
import re
import gc
import json
import time
import shutil
import sqlite3
import zipfile
import warnings
import tracemalloc
from pathlib import Path
from collections import Counter

warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords, wordnet
from nltk.stem import PorterStemmer, SnowballStemmer, WordNetLemmatizer
from nltk import pos_tag

import spacy

try:
    import emoji
except ImportError:
    emoji = None

try:
    import contractions
except ImportError:
    contractions = None

try:
    from wordcloud import WordCloud
except ImportError:
    WordCloud = None

try:
    from matplotlib_venn import venn2
except ImportError:
    venn2 = None


# =============================================================================
# 1. PATH CONFIGURATION
# =============================================================================

PROJECT_DIR = Path(
    r"C:\Users\sagni\Downloads\Text Preprocessing Pipelines using NLTK and spaCy"
)

PROJECT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# -------------------------------------------------------------------------
# Search for dataset in:
# 1. New project folder
# 2. Previous project folder
# -------------------------------------------------------------------------

NEW_CSV = (
    PROJECT_DIR
    / "Tweets.csv"
)

OLD_CSV = Path(
    r"C:\Users\sagni\Downloads\Text Preprocessing Pipeline - NLTK vs spaCy\Tweets.csv"
)

NEW_SQLITE = (
    PROJECT_DIR
    / "database.sqlite"
)

OLD_SQLITE = Path(
    r"C:\Users\sagni\Downloads\Text Preprocessing Pipeline - NLTK vs spaCy\database.sqlite"
)


MAX_TWEETS = 10000
RANDOM_STATE = 42

np.random.seed(
    RANDOM_STATE
)


print("=" * 100)
print("TEXT PREPROCESSING PIPELINES USING NLTK AND spaCy")
print("SLIM SUBMISSION VERSION")
print("=" * 100)

print("\nResults folder:")
print(PROJECT_DIR)


# =============================================================================
# 2. NLTK DOWNLOADS
# =============================================================================

print("\nChecking NLTK resources...")


NLTK_RESOURCES = [
    "punkt",
    "punkt_tab",
    "stopwords",
    "wordnet",
    "omw-1.4",
    "averaged_perceptron_tagger",
    "averaged_perceptron_tagger_eng",
]


for resource in NLTK_RESOURCES:

    try:

        nltk.download(
            resource,
            quiet=True
        )

        print(
            "[OK]",
            resource
        )

    except Exception as e:

        print(
            "[WARNING]",
            resource,
            e
        )


# =============================================================================
# 3. LOAD spaCy
# =============================================================================

print("\nLoading spaCy model...")


try:

    nlp = spacy.load(
        "en_core_web_sm"
    )

except OSError:

    raise RuntimeError(
        "\nspaCy model en_core_web_sm is not installed.\n\n"
        "Run:\n"
        "python -m spacy download en_core_web_sm\n"
    )


print(
    "[OK] en_core_web_sm loaded"
)


# =============================================================================
# 4. LOAD DATASET
# =============================================================================

print("\n" + "=" * 100)
print("LOADING DATASET")
print("=" * 100)


if NEW_CSV.exists():

    DATA_PATH = NEW_CSV

elif OLD_CSV.exists():

    DATA_PATH = OLD_CSV

else:

    DATA_PATH = None


if DATA_PATH is not None:

    print(
        "Using CSV:"
    )

    print(
        DATA_PATH
    )

    df = pd.read_csv(
        DATA_PATH,
        encoding="utf-8",
        on_bad_lines="skip"
    )


else:

    if NEW_SQLITE.exists():

        SQLITE_PATH = NEW_SQLITE

    elif OLD_SQLITE.exists():

        SQLITE_PATH = OLD_SQLITE

    else:

        raise FileNotFoundError(
            "\nCould not find Tweets.csv or database.sqlite.\n"
        )


    print(
        "Using SQLite:"
    )

    print(
        SQLITE_PATH
    )


    connection = sqlite3.connect(
        SQLITE_PATH
    )


    tables = pd.read_sql_query(
        "SELECT name FROM sqlite_master WHERE type='table';",
        connection
    )


    if tables.empty:

        connection.close()

        raise ValueError(
            "No tables found in database.sqlite"
        )


    print(
        "\nTables:"
    )

    print(
        tables
    )


    table_name = (
        tables.iloc[0]["name"]
    )


    df = pd.read_sql_query(
        'SELECT * FROM "'
        + str(table_name)
        + '"',
        connection
    )


    connection.close()


print(
    "\nOriginal dataset shape:",
    df.shape
)


print(
    "\nColumns:"
)

print(
    df.columns.tolist()
)


# =============================================================================
# 5. DETECT TEXT AND SENTIMENT
# =============================================================================

TEXT_COLUMN = None


for candidate in [
    "text",
    "tweet",
    "content",
    "comment",
    "review"
]:

    if candidate in df.columns:

        TEXT_COLUMN = candidate

        break


if TEXT_COLUMN is None:

    object_columns = (
        df.select_dtypes(
            include=["object"]
        )
        .columns
        .tolist()
    )


    if not object_columns:

        raise ValueError(
            "No text column found."
        )


    TEXT_COLUMN = (
        object_columns[-1]
    )


SENTIMENT_COLUMN = None


for candidate in [
    "airline_sentiment",
    "sentiment",
    "label",
    "target",
    "class"
]:

    if candidate in df.columns:

        SENTIMENT_COLUMN = candidate

        break


print(
    "\nText column:",
    TEXT_COLUMN
)


print(
    "Sentiment column:",
    SENTIMENT_COLUMN
)


# =============================================================================
# 6. DATASET CLEANUP
# =============================================================================

df = df.dropna(
    subset=[TEXT_COLUMN]
).copy()


df[TEXT_COLUMN] = (
    df[TEXT_COLUMN]
    .astype(str)
    .str.strip()
)


df = df[
    df[TEXT_COLUMN] != ""
].copy()


# =============================================================================
# 7. SAMPLE 10,000
# =============================================================================

if len(df) > MAX_TWEETS:

    df = df.sample(
        n=MAX_TWEETS,
        random_state=RANDOM_STATE
    )


df = df.reset_index(
    drop=True
)


print(
    "\nTweets used:",
    len(df)
)


# =============================================================================
# 8. MISSING VALUES
# =============================================================================

missing_df = pd.DataFrame({

    "Column":
        df.columns,

    "Missing_Count":
        df.isnull().sum().values,

    "Missing_Percentage":
        (
            df.isnull().mean().values
            * 100
        )

})


missing_df.to_csv(
    PROJECT_DIR
    / "missing_values.csv",
    index=False
)


print(
    "\nMissing values:"
)

print(
    missing_df
)


# =============================================================================
# 9. SAMPLE TWEETS
# =============================================================================

sample_rows = []


if SENTIMENT_COLUMN is not None:

    for sentiment in (
        df[SENTIMENT_COLUMN]
        .dropna()
        .unique()
    ):

        print(
            "\nSentiment:",
            sentiment
        )


        sample = df[
            df[SENTIMENT_COLUMN]
            == sentiment
        ].head(5)


        for _, row in sample.iterrows():

            print(
                "-",
                row[TEXT_COLUMN]
            )


            sample_rows.append({

                "Sentiment":
                    sentiment,

                "Tweet":
                    row[TEXT_COLUMN]

            })


pd.DataFrame(
    sample_rows
).to_csv(
    PROJECT_DIR
    / "sample_tweets.csv",
    index=False
)


# =============================================================================
# 10. PATTERNS
# =============================================================================

URL_PATTERN = (
    r"https?://\S+|www\.\S+"
)

MENTION_PATTERN = (
    r"@\w+"
)

HASHTAG_PATTERN = (
    r"#\w+"
)


# =============================================================================
# 11. INITIAL EDA
# =============================================================================

df[
    "original_char_length"
] = (
    df[TEXT_COLUMN]
    .str.len()
)


df[
    "original_word_count"
] = (
    df[TEXT_COLUMN]
    .str.split()
    .str.len()
)


df[
    "hashtag_count"
] = (
    df[TEXT_COLUMN]
    .str.count(
        HASHTAG_PATTERN
    )
)


df[
    "mention_count"
] = (
    df[TEXT_COLUMN]
    .str.count(
        MENTION_PATTERN
    )
)


df[
    "url_count"
] = (
    df[TEXT_COLUMN]
    .str.count(
        URL_PATTERN
    )
)


statistics_df = pd.DataFrame({

    "Metric": [
        "Tweets",
        "Hashtags",
        "Mentions",
        "URLs",
        "Average Characters",
        "Average Words"
    ],

    "Value": [
        len(df),

        int(
            df[
                "hashtag_count"
            ].sum()
        ),

        int(
            df[
                "mention_count"
            ].sum()
        ),

        int(
            df[
                "url_count"
            ].sum()
        ),

        round(
            df[
                "original_char_length"
            ].mean(),
            3
        ),

        round(
            df[
                "original_word_count"
            ].mean(),
            3
        )
    ]

})


statistics_df.to_csv(
    PROJECT_DIR
    / "social_media_statistics.csv",
    index=False
)


print(
    "\nDataset statistics:"
)

print(
    statistics_df
)


# =============================================================================
# 12. GRAPH SAVER
# =============================================================================

def save_graph(
    filename
):

    plt.tight_layout()

    plt.savefig(
        PROJECT_DIR
        / filename,
        dpi=130,
        bbox_inches="tight",
        optimize=True
        if filename.lower().endswith(
            ".jpg"
        )
        else False
    )

    plt.show()

    plt.close()


# =============================================================================
# 13. TWEET LENGTH DISTRIBUTION
# =============================================================================

plt.figure(
    figsize=(9, 5)
)


sns.histplot(
    df[
        "original_char_length"
    ],
    bins=40,
    kde=True
)


plt.title(
    "Tweet Length Distribution"
)

plt.xlabel(
    "Characters"
)

plt.ylabel(
    "Frequency"
)


# save directly because savefig PNG has no optimize argument
plt.tight_layout()

plt.savefig(
    PROJECT_DIR
    / "tweet_length_distribution.png",
    dpi=130,
    bbox_inches="tight"
)

plt.show()

plt.close()


# =============================================================================
# 14. CONTRACTION HANDLING
# =============================================================================

FALLBACK_CONTRACTIONS = {

    "can't":
        "cannot",

    "won't":
        "will not",

    "don't":
        "do not",

    "doesn't":
        "does not",

    "didn't":
        "did not",

    "isn't":
        "is not",

    "aren't":
        "are not",

    "wasn't":
        "was not",

    "weren't":
        "were not",

    "couldn't":
        "could not",

    "wouldn't":
        "would not",

    "shouldn't":
        "should not",

    "haven't":
        "have not",

    "hasn't":
        "has not",

    "hadn't":
        "had not",

    "i'm":
        "i am",

    "i've":
        "i have",

    "i'll":
        "i will",

    "you're":
        "you are",

    "we're":
        "we are",

    "they're":
        "they are",

    "it's":
        "it is",

    "that's":
        "that is"

}


def expand_contractions(
    text
):

    if contractions is not None:

        try:

            return contractions.fix(
                text
            )

        except Exception:

            pass


    for short, full in (
        FALLBACK_CONTRACTIONS.items()
    ):

        text = re.sub(
            re.escape(short),
            full,
            text,
            flags=re.IGNORECASE
        )


    return text


# =============================================================================
# 15. REPEATED CHARACTERS
# =============================================================================

def normalize_repeated(
    text
):

    return re.sub(
        r"(.)\1{2,}",
        r"\1\1",
        text
    )


# =============================================================================
# 16. EMOJI
# =============================================================================

def handle_emoji(
    text,
    mode="text"
):

    if emoji is None:

        return text


    if mode == "remove":

        return emoji.replace_emoji(
            text,
            replace=""
        )


    if mode == "text":

        return emoji.demojize(
            text,
            delimiters=(
                " ",
                " "
            )
        )


    return text


# =============================================================================
# 17. CLEAN TEXT
# =============================================================================

def clean_text(
    text,
    keep_hashtags=True,
    keep_numbers=False,
    emoji_mode="text"
):

    if pd.isna(text):

        return ""


    text = str(
        text
    ).lower()


    text = handle_emoji(
        text,
        emoji_mode
    )


    text = expand_contractions(
        text
    )


    text = re.sub(
        URL_PATTERN,
        " ",
        text
    )


    text = re.sub(
        MENTION_PATTERN,
        " ",
        text
    )


    if keep_hashtags:

        text = re.sub(
            r"#(\w+)",
            r"\1",
            text
        )

    else:

        text = re.sub(
            HASHTAG_PATTERN,
            " ",
            text
        )


    text = normalize_repeated(
        text
    )


    if not keep_numbers:

        text = re.sub(
            r"\b\d+(?:\.\d+)?\b",
            " ",
            text
        )


    text = text.replace(
        "_",
        " "
    )


    text = re.sub(
        r"[^a-z0-9\s]",
        " ",
        text
    )


    text = re.sub(
        r"\s+",
        " ",
        text
    ).strip()


    return text


# =============================================================================
# 18. CLEAN DATASET
# =============================================================================

print(
    "\nCleaning tweets..."
)


df[
    "clean_text"
] = (
    df[TEXT_COLUMN]
    .apply(
        lambda x:
            clean_text(
                x,
                emoji_mode="text"
            )
    )
)


df[
    "emoji_removed_text"
] = (
    df[TEXT_COLUMN]
    .apply(
        lambda x:
            clean_text(
                x,
                emoji_mode="remove"
            )
    )
)


df[
    "clean_char_length"
] = (
    df[
        "clean_text"
    ].str.len()
)


df[
    "clean_word_count"
] = (
    df[
        "clean_text"
    ]
    .str.split()
    .str.len()
)


# =============================================================================
# 19. BEFORE VS AFTER GRAPH
# =============================================================================

plt.figure(
    figsize=(9, 5)
)


sns.histplot(
    df[
        "original_char_length"
    ],
    bins=40,
    alpha=0.45,
    label="Before"
)


sns.histplot(
    df[
        "clean_char_length"
    ],
    bins=40,
    alpha=0.45,
    label="After"
)


plt.title(
    "Tweet Length Before and After Preprocessing"
)

plt.xlabel(
    "Characters"
)

plt.ylabel(
    "Frequency"
)

plt.legend()


plt.tight_layout()

plt.savefig(
    PROJECT_DIR
    / "before_after_length_distribution.png",
    dpi=130,
    bbox_inches="tight"
)

plt.show()

plt.close()


# =============================================================================
# 20. EMOJI COMPARISON
# =============================================================================

emoji_comparison = pd.DataFrame({

    "Method": [
        "Emoji to Text",
        "Emoji Removed"
    ],

    "Average_Characters": [

        df[
            "clean_text"
        ].str.len().mean(),

        df[
            "emoji_removed_text"
        ].str.len().mean()

    ]

})


emoji_comparison.to_csv(
    PROJECT_DIR
    / "emoji_comparison.csv",
    index=False
)


# =============================================================================
# 21. NLTK SETUP
# =============================================================================

NEGATION_WORDS = {

    "not",
    "no",
    "nor",
    "never"

}


CUSTOM_STOPWORDS = {

    "amp",
    "rt",
    "via",
    "tweet",
    "tweets"

}


NLTK_STOPWORDS = set(
    stopwords.words(
        "english"
    )
)


NLTK_STOPWORDS -= (
    NEGATION_WORDS
)


NLTK_STOPWORDS.update(
    CUSTOM_STOPWORDS
)


porter = PorterStemmer()


snowball = SnowballStemmer(
    "english"
)


lemmatizer = (
    WordNetLemmatizer()
)


def get_wordnet_pos(
    tag
):

    if tag.startswith(
        "J"
    ):

        return wordnet.ADJ


    if tag.startswith(
        "V"
    ):

        return wordnet.VERB


    if tag.startswith(
        "N"
    ):

        return wordnet.NOUN


    if tag.startswith(
        "R"
    ):

        return wordnet.ADV


    return wordnet.NOUN


# =============================================================================
# 22. TOKENIZATION COMPARISON
# =============================================================================

tokenization_rows = []


for text in (
    df[
        "clean_text"
    ].head(20)
):

    nltk_tokens = (
        word_tokenize(
            text
        )
    )


    split_tokens = (
        text.split()
    )


    tokenization_rows.append({

        "Text":
            text,

        "NLTK":
            " | ".join(
                nltk_tokens
            ),

        "Split":
            " | ".join(
                split_tokens
            ),

        "NLTK_Count":
            len(
                nltk_tokens
            ),

        "Split_Count":
            len(
                split_tokens
            )

    })


pd.DataFrame(
    tokenization_rows
).to_csv(
    PROJECT_DIR
    / "tokenization_comparison.csv",
    index=False
)


# =============================================================================
# 23. NLTK PIPELINE
# =============================================================================

def nltk_preprocess(
    text
):

    cleaned = clean_text(
        text
    )


    if not cleaned:

        return {

            "tokens": [],

            "filtered": [],

            "porter": [],

            "snowball": [],

            "lemmas": [],

            "pos": [],

            "nouns_adjectives": []

        }


    tokens = word_tokenize(
        cleaned
    )


    tokens = [

        token

        for token in tokens

        if token.isalpha()

    ]


    filtered = [

        token

        for token in tokens

        if (
            token
            not in NLTK_STOPWORDS

            or

            token
            in NEGATION_WORDS
        )

    ]


    tagged = (

        pos_tag(
            filtered
        )

        if filtered

        else []

    )


    lemmas = [

        lemmatizer.lemmatize(
            token,
            get_wordnet_pos(
                tag
            )
        )

        for token, tag
        in tagged

    ]


    nouns_adjectives = [

        token

        for token, tag
        in tagged

        if (
            tag.startswith(
                "N"
            )

            or

            tag.startswith(
                "J"
            )
        )

    ]


    return {

        "tokens":
            tokens,

        "filtered":
            filtered,

        "porter":
            [
                porter.stem(
                    x
                )
                for x
                in filtered
            ],

        "snowball":
            [
                snowball.stem(
                    x
                )
                for x
                in filtered
            ],

        "lemmas":
            lemmas,

        "pos":
            tagged,

        "nouns_adjectives":
            nouns_adjectives

    }


# =============================================================================
# 24. RUN NLTK
# =============================================================================

print(
    "\nRunning NLTK..."
)


nltk_results = []


start = (
    time.perf_counter()
)


for i, text in enumerate(
    df[TEXT_COLUMN],
    1
):

    nltk_results.append(
        nltk_preprocess(
            text
        )
    )


    if i % 1000 == 0:

        print(
            "NLTK:",
            i,
            "/",
            len(df)
        )


nltk_full_time = (
    time.perf_counter()
    - start
)


print(
    "NLTK finished in",
    round(
        nltk_full_time,
        3
    ),
    "seconds"
)


# =============================================================================
# 25. NLTK RESULTS
# =============================================================================

nltk_df = pd.DataFrame({

    "original_text":
        df[
            TEXT_COLUMN
        ].values,

    "clean_text":
        df[
            "clean_text"
        ].values,

    "tokens":
        [
            " ".join(
                x["tokens"]
            )
            for x
            in nltk_results
        ],

    "filtered_tokens":
        [
            " ".join(
                x["filtered"]
            )
            for x
            in nltk_results
        ],

    "porter_stems":
        [
            " ".join(
                x["porter"]
            )
            for x
            in nltk_results
        ],

    "snowball_stems":
        [
            " ".join(
                x["snowball"]
            )
            for x
            in nltk_results
        ],

    "lemmas":
        [
            " ".join(
                x["lemmas"]
            )
            for x
            in nltk_results
        ],

    "token_count":
        [
            len(
                x["filtered"]
            )
            for x
            in nltk_results
        ]

})


if SENTIMENT_COLUMN:

    nltk_df[
        "sentiment"
    ] = (
        df[
            SENTIMENT_COLUMN
        ].values
    )


# Full result stays in folder.
# It will NOT go into the slim ZIP.

nltk_df.to_csv(
    PROJECT_DIR
    / "nltk_processed_tweets.csv",
    index=False
)


# =============================================================================
# 26. NLTK POS
# =============================================================================

nltk_pos_counter = (
    Counter()
)


for result in nltk_results:

    for _, tag in (
        result["pos"]
    ):

        nltk_pos_counter[
            tag
        ] += 1


nltk_pos_df = pd.DataFrame(
    nltk_pos_counter.items(),
    columns=[
        "POS",
        "Count"
    ]
).sort_values(
    "Count",
    ascending=False
)


nltk_pos_df.to_csv(
    PROJECT_DIR
    / "nltk_pos_distribution.csv",
    index=False
)


if not nltk_pos_df.empty:

    top = (
        nltk_pos_df
        .head(10)
    )


    plt.figure(
        figsize=(8, 8)
    )


    plt.pie(
        top["Count"],
        labels=top["POS"],
        autopct="%1.1f%%",
        startangle=90
    )


    plt.title(
        "NLTK POS Distribution"
    )


    plt.tight_layout()

    plt.savefig(
        PROJECT_DIR
        / "nltk_pos_distribution.png",
        dpi=120,
        bbox_inches="tight"
    )

    plt.show()

    plt.close()


# =============================================================================
# 27. STEMMING VS LEMMATIZATION
# =============================================================================

comparison_words = [

    "running",
    "runs",
    "ran",
    "better",
    "good",
    "best",
    "studies",
    "studying",
    "studied",
    "flies",
    "flying",
    "happier",
    "happiness"

]


stem_rows = []


for word in comparison_words:

    tag = pos_tag(
        [word]
    )[0][1]


    lemma = (
        lemmatizer.lemmatize(
            word,
            get_wordnet_pos(
                tag
            )
        )
    )


    stem_rows.append({

        "Word":
            word,

        "POS":
            tag,

        "Porter":
            porter.stem(
                word
            ),

        "Snowball":
            snowball.stem(
                word
            ),

        "Lemma":
            lemma

    })


stem_df = pd.DataFrame(
    stem_rows
)


stem_df.to_csv(
    PROJECT_DIR
    / "stemming_vs_lemmatization.csv",
    index=False
)


modified_df = pd.DataFrame({

    "Method": [
        "Porter",
        "Snowball",
        "Lemmatizer"
    ],

    "Words_Modified": [

        int(
            (
                stem_df[
                    "Word"
                ]
                !=
                stem_df[
                    "Porter"
                ]
            ).sum()
        ),

        int(
            (
                stem_df[
                    "Word"
                ]
                !=
                stem_df[
                    "Snowball"
                ]
            ).sum()
        ),

        int(
            (
                stem_df[
                    "Word"
                ]
                !=
                stem_df[
                    "Lemma"
                ]
            ).sum()
        )

    ]

})


plt.figure(
    figsize=(8, 5)
)


sns.barplot(
    data=modified_df,
    x="Method",
    y="Words_Modified"
)


plt.title(
    "Stemming vs Lemmatization"
)


plt.tight_layout()

plt.savefig(
    PROJECT_DIR
    / "stemming_lemmatization_comparison.png",
    dpi=120,
    bbox_inches="tight"
)

plt.show()

plt.close()


# =============================================================================
# 28. STOPWORD IMPACT
# =============================================================================

before_stop = [

    len(
        x["tokens"]
    )

    for x
    in nltk_results

]


after_stop = [

    len(
        x["filtered"]
    )

    for x
    in nltk_results

]


stopword_df = pd.DataFrame({

    "Stage": [
        "Before",
        "After"
    ],

    "Average_Tokens": [
        np.mean(
            before_stop
        ),
        np.mean(
            after_stop
        )
    ]

})


plt.figure(
    figsize=(7, 5)
)


sns.barplot(
    data=stopword_df,
    x="Stage",
    y="Average_Tokens"
)


plt.title(
    "Stopword Removal Impact"
)


plt.ylabel(
    "Average Tokens"
)


plt.tight_layout()

plt.savefig(
    PROJECT_DIR
    / "stopword_removal_impact.png",
    dpi=120,
    bbox_inches="tight"
)

plt.show()

plt.close()


# =============================================================================
# 29. spaCy CUSTOMIZATION
# =============================================================================

for word in CUSTOM_STOPWORDS:

    nlp.vocab[
        word
    ].is_stop = True


for word in NEGATION_WORDS:

    nlp.vocab[
        word
    ].is_stop = False


# =============================================================================
# 30. CUSTOM ENTITY RULER
# =============================================================================

if (
    "custom_entity_ruler"
    not in nlp.pipe_names
):

    ruler = nlp.add_pipe(
        "entity_ruler",
        name="custom_entity_ruler",
        before="ner"
    )


    ruler.add_patterns([

        {
            "label":
                "AIRLINE",

            "pattern": [
                {
                    "LOWER":
                        "united"
                }
            ]
        },

        {
            "label":
                "AIRLINE",

            "pattern": [
                {
                    "LOWER":
                        "delta"
                }
            ]
        },

        {
            "label":
                "AIRLINE",

            "pattern": [
                {
                    "LOWER":
                        "southwest"
                }
            ]
        },

        {
            "label":
                "AIRLINE",

            "pattern": [
                {
                    "LOWER":
                        "american"
                },
                {
                    "LOWER":
                        "airlines"
                }
            ]
        },

        {
            "label":
                "AIRLINE",

            "pattern": [
                {
                    "LOWER":
                        "us"
                },
                {
                    "LOWER":
                        "airways"
                }
            ]
        },

        {
            "label":
                "AIRLINE",

            "pattern": [
                {
                    "LOWER":
                        "virgin"
                },
                {
                    "LOWER":
                        "america"
                }
            ]
        }

    ])


# =============================================================================
# 31. spaCy PROCESSOR
# =============================================================================

def process_spacy_doc(
    doc
):

    tokens = []

    lemmas = []

    pos_tags = []

    entities = []


    for token in doc:

        if (
            token.is_space
            or
            token.is_punct
            or
            not token.is_alpha
        ):

            continue


        lower = (
            token.text.lower()
        )


        tokens.append(
            lower
        )


        pos_tags.append(
            (
                token.text,
                token.pos_
            )
        )


        if (
            not token.is_stop

            or

            lower
            in NEGATION_WORDS
        ):

            lemma = (
                token.lemma_.lower()
            )


            if lemma:

                lemmas.append(
                    lemma
                )


    for ent in doc.ents:

        entities.append({

            "text":
                ent.text,

            "label":
                ent.label_

        })


    return {

        "tokens":
            tokens,

        "lemmas":
            lemmas,

        "pos":
            pos_tags,

        "entities":
            entities

    }


# =============================================================================
# 32. RUN spaCy
# =============================================================================

print(
    "\nRunning spaCy..."
)


clean_texts = (
    df[
        "clean_text"
    ]
    .fillna("")
    .astype(str)
    .tolist()
)


spacy_results = []


start = (
    time.perf_counter()
)


for i, doc in enumerate(
    nlp.pipe(
        clean_texts,
        batch_size=128
    ),
    1
):

    spacy_results.append(
        process_spacy_doc(
            doc
        )
    )


    if i % 1000 == 0:

        print(
            "spaCy:",
            i,
            "/",
            len(df)
        )


spacy_full_time = (
    time.perf_counter()
    - start
)


print(
    "spaCy finished in",
    round(
        spacy_full_time,
        3
    ),
    "seconds"
)


# =============================================================================
# 33. spaCy RESULTS
# =============================================================================

spacy_df = pd.DataFrame({

    "original_text":
        df[
            TEXT_COLUMN
        ].values,

    "clean_text":
        df[
            "clean_text"
        ].values,

    "tokens":
        [
            " ".join(
                x["tokens"]
            )
            for x
            in spacy_results
        ],

    "lemmas":
        [
            " ".join(
                x["lemmas"]
            )
            for x
            in spacy_results
        ],

    "token_count":
        [
            len(
                x["lemmas"]
            )
            for x
            in spacy_results
        ]

})


if SENTIMENT_COLUMN:

    spacy_df[
        "sentiment"
    ] = (
        df[
            SENTIMENT_COLUMN
        ].values
    )


spacy_df.to_csv(
    PROJECT_DIR
    / "spacy_processed_tweets.csv",
    index=False
)


# =============================================================================
# 34. spaCy POS
# =============================================================================

spacy_pos_counter = (
    Counter()
)


for result in spacy_results:

    for _, tag in (
        result["pos"]
    ):

        spacy_pos_counter[
            tag
        ] += 1


spacy_pos_df = pd.DataFrame(
    spacy_pos_counter.items(),
    columns=[
        "POS",
        "Count"
    ]
).sort_values(
    "Count",
    ascending=False
)


spacy_pos_df.to_csv(
    PROJECT_DIR
    / "spacy_pos_distribution.csv",
    index=False
)


if not spacy_pos_df.empty:

    top = (
        spacy_pos_df
        .head(10)
    )


    plt.figure(
        figsize=(8, 8)
    )


    plt.pie(
        top["Count"],
        labels=top["POS"],
        autopct="%1.1f%%",
        startangle=90
    )


    plt.title(
        "spaCy POS Distribution"
    )


    plt.tight_layout()

    plt.savefig(
        PROJECT_DIR
        / "spacy_pos_distribution.png",
        dpi=120,
        bbox_inches="tight"
    )

    plt.show()

    plt.close()


# =============================================================================
# 35. ENTITIES
# =============================================================================

all_entities = []

entity_counter = (
    Counter()
)


for index, result in enumerate(
    spacy_results
):

    for entity in (
        result[
            "entities"
        ]
    ):

        all_entities.append({

            "tweet_index":
                int(index),

            "entity":
                entity[
                    "text"
                ],

            "label":
                entity[
                    "label"
                ]

        })


        entity_counter[
            entity[
                "label"
            ]
        ] += 1


# Full JSON stays outside slim submission ZIP

with open(
    PROJECT_DIR
    / "extracted_entities.json",
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        all_entities,
        f,
        ensure_ascii=False,
        indent=2
    )


entity_frequency_df = pd.DataFrame(
    entity_counter.items(),
    columns=[
        "Entity_Type",
        "Count"
    ]
).sort_values(
    "Count",
    ascending=False
)


entity_frequency_df.to_csv(
    PROJECT_DIR
    / "entity_frequency.csv",
    index=False
)


if not entity_frequency_df.empty:

    plt.figure(
        figsize=(9, 6)
    )


    sns.barplot(
        data=(
            entity_frequency_df
            .head(15)
        ),
        y="Entity_Type",
        x="Count"
    )


    plt.title(
        "Named Entity Frequency"
    )


    plt.tight_layout()

    plt.savefig(
        PROJECT_DIR
        / "named_entity_frequency.png",
        dpi=120,
        bbox_inches="tight"
    )

    plt.show()

    plt.close()


# =============================================================================
# 36. TOKEN COUNT DISTRIBUTION
# =============================================================================

plt.figure(
    figsize=(9, 5)
)


sns.histplot(
    nltk_df[
        "token_count"
    ],
    bins=30,
    alpha=0.45,
    label="NLTK"
)


sns.histplot(
    spacy_df[
        "token_count"
    ],
    bins=30,
    alpha=0.45,
    label="spaCy"
)


plt.title(
    "Token Count Distribution"
)

plt.xlabel(
    "Tokens per Tweet"
)

plt.ylabel(
    "Frequency"
)

plt.legend()


plt.tight_layout()

plt.savefig(
    PROJECT_DIR
    / "token_count_distribution.png",
    dpi=120,
    bbox_inches="tight"
)

plt.show()

plt.close()


# =============================================================================
# 37. VOCABULARY
# =============================================================================

original_vocabulary = set()


for text in df[
    TEXT_COLUMN
]:

    original_vocabulary.update(
        re.findall(
            r"\b[a-zA-Z]+\b",
            str(
                text
            ).lower()
        )
    )


nltk_vocabulary = set()


for result in nltk_results:

    nltk_vocabulary.update(
        result[
            "lemmas"
        ]
    )


spacy_vocabulary = set()


for result in spacy_results:

    spacy_vocabulary.update(
        result[
            "lemmas"
        ]
    )


vocabulary_df = pd.DataFrame({

    "Pipeline": [
        "Original",
        "NLTK",
        "spaCy"
    ],

    "Unique_Tokens": [
        len(
            original_vocabulary
        ),
        len(
            nltk_vocabulary
        ),
        len(
            spacy_vocabulary
        )
    ]

})


vocabulary_df.to_csv(
    PROJECT_DIR
    / "vocabulary_statistics.csv",
    index=False
)


# =============================================================================
# 38. VOCABULARY GRAPH
# =============================================================================

plt.figure(
    figsize=(7, 5)
)


sns.barplot(
    data=vocabulary_df,
    x="Pipeline",
    y="Unique_Tokens"
)


plt.title(
    "Vocabulary Size Comparison"
)


plt.tight_layout()

plt.savefig(
    PROJECT_DIR
    / "vocabulary_size_comparison.png",
    dpi=120,
    bbox_inches="tight"
)

plt.show()

plt.close()


# =============================================================================
# 39. VOCABULARY VENN
# =============================================================================

if venn2 is not None:

    plt.figure(
        figsize=(7, 7)
    )


    venn2(
        [
            nltk_vocabulary,
            spacy_vocabulary
        ],
        set_labels=(
            "NLTK",
            "spaCy"
        )
    )


    plt.title(
        "Vocabulary Overlap"
    )


    plt.tight_layout()

    plt.savefig(
        PROJECT_DIR
        / "vocabulary_overlap.png",
        dpi=120,
        bbox_inches="tight"
    )

    plt.show()

    plt.close()


# =============================================================================
# 40. WORD CLOUDS
# =============================================================================

if WordCloud is not None:

    original_cloud_text = (
        " ".join(
            df[
                TEXT_COLUMN
            ]
            .astype(str)
            .tolist()
        )
    )


    if original_cloud_text.strip():

        cloud = WordCloud(
            width=900,
            height=450,
            background_color="white",
            max_words=150
        ).generate(
            original_cloud_text
        )


        plt.figure(
            figsize=(10, 5)
        )

        plt.imshow(
            cloud,
            interpolation="bilinear"
        )

        plt.axis(
            "off"
        )

        plt.title(
            "Word Cloud Before Preprocessing"
        )

        plt.tight_layout()

        plt.savefig(
            PROJECT_DIR
            / "wordcloud_before.png",
            dpi=100,
            bbox_inches="tight"
        )

        plt.show()

        plt.close()


    processed_cloud_text = (
        " ".join(
            nltk_df[
                "lemmas"
            ]
            .astype(str)
            .tolist()
        )
    )


    if processed_cloud_text.strip():

        cloud = WordCloud(
            width=900,
            height=450,
            background_color="white",
            max_words=150
        ).generate(
            processed_cloud_text
        )


        plt.figure(
            figsize=(10, 5)
        )

        plt.imshow(
            cloud,
            interpolation="bilinear"
        )

        plt.axis(
            "off"
        )

        plt.title(
            "Word Cloud After Preprocessing"
        )

        plt.tight_layout()

        plt.savefig(
            PROJECT_DIR
            / "wordcloud_after.png",
            dpi=100,
            bbox_inches="tight"
        )

        plt.show()

        plt.close()


# =============================================================================
# 41. DEPENDENCY PARSE EXAMPLE
# =============================================================================

dependency_text = ""


for text in clean_texts:

    words = (
        text.split()
    )


    if (
        len(words) >= 6
        and
        len(words) <= 14
    ):

        dependency_text = text

        break


if not dependency_text:

    dependency_text = (
        "the airline cancelled my flight today"
    )


dependency_doc = nlp(
    dependency_text
)


dependency_rows = []


for token in dependency_doc:

    dependency_rows.append({

        "Token":
            token.text,

        "POS":
            token.pos_,

        "Dependency":
            token.dep_,

        "Head":
            token.head.text

    })


dependency_df = pd.DataFrame(
    dependency_rows
)


dependency_df.to_csv(
    PROJECT_DIR
    / "dependency_parse_example.csv",
    index=False
)


# Simple dependency visualization

tokens = [
    x.text
    for x
    in dependency_doc
]


heads = [
    x.head.i
    for x
    in dependency_doc
]


deps = [
    x.dep_
    for x
    in dependency_doc
]


plt.figure(
    figsize=(
        max(
            10,
            len(tokens)
        ),
        5
    )
)


ax = plt.gca()


for i, token in enumerate(
    tokens
):

    ax.text(
        i,
        0,
        token,
        ha="center",
        va="center",
        fontsize=9,
        bbox={
            "boxstyle":
                "round,pad=0.3",
            "fill":
                False
        }
    )


for i, head in enumerate(
    heads
):

    if i == head:

        continue


    rad = (
        0.25
        if head > i
        else -0.25
    )


    ax.annotate(
        "",
        xy=(
            head,
            0.15
        ),
        xytext=(
            i,
            0.15
        ),
        arrowprops={
            "arrowstyle":
                "->",
            "connectionstyle":
                "arc3,rad="
                + str(rad)
        }
    )


    ax.text(
        (
            i
            + head
        )
        / 2,
        0.7
        + abs(
            head
            - i
        )
        * 0.12,
        deps[i],
        fontsize=7,
        ha="center"
    )


ax.set_xlim(
    -1,
    len(tokens)
)


ax.set_ylim(
    -1,
    4
)


ax.axis(
    "off"
)


plt.title(
    "spaCy Dependency Parse\n"
    + dependency_text
)


plt.tight_layout()

plt.savefig(
    PROJECT_DIR
    / "dependency_parse_example.png",
    dpi=120,
    bbox_inches="tight"
)

plt.show()

plt.close()


# =============================================================================
# 42. BENCHMARK FUNCTIONS
# =============================================================================

def benchmark_nltk(
    texts
):

    gc.collect()


    tracemalloc.start()


    start = (
        time.perf_counter()
    )


    for text in texts:

        nltk_preprocess(
            text
        )


    elapsed = (
        time.perf_counter()
        - start
    )


    _, peak = (
        tracemalloc
        .get_traced_memory()
    )


    tracemalloc.stop()


    memory = (
        peak
        / 1024
        / 1024
    )


    speed = (
        len(texts)
        / elapsed
        if elapsed > 0
        else 0
    )


    return (
        elapsed,
        speed,
        memory
    )


def benchmark_spacy(
    texts
):

    gc.collect()


    cleaned = [

        clean_text(
            x
        )

        for x
        in texts

    ]


    tracemalloc.start()


    start = (
        time.perf_counter()
    )


    for doc in nlp.pipe(
        cleaned,
        batch_size=128
    ):

        process_spacy_doc(
            doc
        )


    elapsed = (
        time.perf_counter()
        - start
    )


    _, peak = (
        tracemalloc
        .get_traced_memory()
    )


    tracemalloc.stop()


    memory = (
        peak
        / 1024
        / 1024
    )


    speed = (
        len(texts)
        / elapsed
        if elapsed > 0
        else 0
    )


    return (
        elapsed,
        speed,
        memory
    )


# =============================================================================
# 43. RUN BENCHMARK
# =============================================================================

print(
    "\n" + "=" * 100
)

print(
    "BENCHMARKING"
)

print(
    "=" * 100
)


benchmark_rows = []


for size in [
    1000,
    5000,
    10000
]:

    actual_size = min(
        size,
        len(df)
    )


    texts = (
        df[
            TEXT_COLUMN
        ]
        .head(
            actual_size
        )
        .astype(str)
        .tolist()
    )


    print(
        "\nTweets:",
        actual_size
    )


    nt, ns, nm = (
        benchmark_nltk(
            texts
        )
    )


    benchmark_rows.append({

        "Library":
            "NLTK",

        "Dataset_Size":
            actual_size,

        "Processing_Time_Seconds":
            nt,

        "Tweets_Per_Second":
            ns,

        "Peak_Memory_MB":
            nm

    })


    print(
        "NLTK:",
        round(
            nt,
            3
        ),
        "sec"
    )


    st, ss, sm = (
        benchmark_spacy(
            texts
        )
    )


    benchmark_rows.append({

        "Library":
            "spaCy",

        "Dataset_Size":
            actual_size,

        "Processing_Time_Seconds":
            st,

        "Tweets_Per_Second":
            ss,

        "Peak_Memory_MB":
            sm

    })


    print(
        "spaCy:",
        round(
            st,
            3
        ),
        "sec"
    )


benchmark_df = pd.DataFrame(
    benchmark_rows
)


benchmark_df.to_csv(
    PROJECT_DIR
    / "benchmark_results.csv",
    index=False
)


# =============================================================================
# 44. BENCHMARK GRAPHS
# =============================================================================

plt.figure(
    figsize=(9, 5)
)


sns.barplot(
    data=benchmark_df,
    x="Dataset_Size",
    y="Processing_Time_Seconds",
    hue="Library"
)


plt.title(
    "Processing Time Comparison"
)


plt.ylabel(
    "Seconds"
)


plt.tight_layout()

plt.savefig(
    PROJECT_DIR
    / "processing_time_comparison.png",
    dpi=120,
    bbox_inches="tight"
)

plt.show()

plt.close()


plt.figure(
    figsize=(9, 5)
)


sns.lineplot(
    data=benchmark_df,
    x="Dataset_Size",
    y="Processing_Time_Seconds",
    hue="Library",
    marker="o"
)


plt.title(
    "Processing Time vs Dataset Size"
)


plt.tight_layout()

plt.savefig(
    PROJECT_DIR
    / "processing_time_scaling.png",
    dpi=120,
    bbox_inches="tight"
)

plt.show()

plt.close()


plt.figure(
    figsize=(9, 5)
)


sns.barplot(
    data=benchmark_df,
    x="Dataset_Size",
    y="Tweets_Per_Second",
    hue="Library"
)


plt.title(
    "Tweets per Second"
)


plt.tight_layout()

plt.savefig(
    PROJECT_DIR
    / "tweets_per_second.png",
    dpi=120,
    bbox_inches="tight"
)

plt.show()

plt.close()


plt.figure(
    figsize=(9, 5)
)


sns.lineplot(
    data=benchmark_df,
    x="Dataset_Size",
    y="Peak_Memory_MB",
    hue="Library",
    marker="o"
)


plt.title(
    "Memory Usage Comparison"
)


plt.ylabel(
    "Peak Memory (MB)"
)


plt.tight_layout()

plt.savefig(
    PROJECT_DIR
    / "memory_usage_comparison.png",
    dpi=120,
    bbox_inches="tight"
)

plt.show()

plt.close()


# =============================================================================
# 45. COMPARISON TABLE
# =============================================================================

largest_size = (
    benchmark_df[
        "Dataset_Size"
    ].max()
)


largest_df = (
    benchmark_df[
        benchmark_df[
            "Dataset_Size"
        ]
        == largest_size
    ]
)


def get_metric(
    library,
    metric
):

    values = (
        largest_df[
            largest_df[
                "Library"
            ]
            == library
        ][metric]
    )


    if values.empty:

        return np.nan


    return float(
        values.iloc[0]
    )


nltk_time = get_metric(
    "NLTK",
    "Processing_Time_Seconds"
)


spacy_time = get_metric(
    "spaCy",
    "Processing_Time_Seconds"
)


nltk_memory = get_metric(
    "NLTK",
    "Peak_Memory_MB"
)


spacy_memory = get_metric(
    "spaCy",
    "Peak_Memory_MB"
)


comparison_df = pd.DataFrame({

    "Metric": [

        "Processing time",

        "Average tokens/tweet",

        "Unique vocabulary",

        "Peak memory",

        "Lemmatization",

        "POS",

        "NER",

        "Dependency Parsing",

        "Recommended Use"

    ],

    "NLTK": [

        round(
            nltk_time,
            3
        ),

        round(
            nltk_df[
                "token_count"
            ].mean(),
            2
        ),

        len(
            nltk_vocabulary
        ),

        round(
            nltk_memory,
            2
        ),

        "WordNet + POS",

        "Perceptron",

        "Limited in this pipeline",

        "Not integrated",

        "Education / custom NLP"

    ],

    "spaCy": [

        round(
            spacy_time,
            3
        ),

        round(
            spacy_df[
                "token_count"
            ].mean(),
            2
        ),

        len(
            spacy_vocabulary
        ),

        round(
            spacy_memory,
            2
        ),

        "Model lemmatizer",

        "Integrated",

        "Built-in",

        "Built-in",

        "Production / integrated NLP"

    ]

})


comparison_df.to_csv(
    PROJECT_DIR
    / "preprocessing_comparison.csv",
    index=False
)


print(
    "\nComparison:"
)

print(
    comparison_df.to_string(
        index=False
    )
)


# =============================================================================
# 46. QUALITY REVIEW - 100 SAMPLES
# =============================================================================

review_size = min(
    100,
    len(df)
)


indices = (
    df.sample(
        n=review_size,
        random_state=42
    ).index
)


quality_df = pd.DataFrame({

    "Original_Text":
        df.loc[
            indices,
            TEXT_COLUMN
        ].values,

    "NLTK_Output":
        nltk_df.loc[
            indices,
            "lemmas"
        ].values,

    "spaCy_Output":
        spacy_df.loc[
            indices,
            "lemmas"
        ].values,

    "NLTK_Lemma_Rating_1_to_5":
        "",

    "spaCy_Lemma_Rating_1_to_5":
        "",

    "NLTK_POS_Correct_Y_N":
        "",

    "spaCy_POS_Correct_Y_N":
        "",

    "spaCy_NER_Correct_Y_N":
        "",

    "Notes":
        ""

})


quality_df.to_csv(
    PROJECT_DIR
    / "quality_review_100_samples.csv",
    index=False
)


# =============================================================================
# 47. FINAL FULL RESULTS
# =============================================================================

final_results = pd.DataFrame({

    "original_text":
        df[
            TEXT_COLUMN
        ].values,

    "clean_text":
        df[
            "clean_text"
        ].values,

    "nltk_processed":
        nltk_df[
            "lemmas"
        ].values,

    "spacy_processed":
        spacy_df[
            "lemmas"
        ].values,

    "nltk_token_count":
        nltk_df[
            "token_count"
        ].values,

    "spacy_token_count":
        spacy_df[
            "token_count"
        ].values

})


if SENTIMENT_COLUMN:

    final_results[
        "sentiment"
    ] = (
        df[
            SENTIMENT_COLUMN
        ].values
    )


final_results.to_csv(
    PROJECT_DIR
    / "final_preprocessing_results.csv",
    index=False
)


# =============================================================================
# 48. REUSABLE preprocessing_library.py
# =============================================================================

preprocessing_code = [
    '"""Reusable social-media preprocessing functions."""',
    "",
    "import re",
    "import pandas as pd",
    "",
    "try:",
    "    import emoji",
    "except ImportError:",
    "    emoji = None",
    "",
    "try:",
    "    import contractions",
    "except ImportError:",
    "    contractions = None",
    "",
    'URL_PATTERN = r"https?://\\S+|www\\.\\S+"',
    'MENTION_PATTERN = r"@\\w+"',
    'HASHTAG_PATTERN = r"#\\w+"',
    "",
    "",
    "def clean_text(text, keep_hashtags=True, keep_numbers=False, emoji_mode='text'):",
    "",
    "    if pd.isna(text):",
    "        return ''",
    "",
    "    text = str(text).lower()",
    "",
    "    if emoji is not None:",
    "        if emoji_mode == 'remove':",
    "            text = emoji.replace_emoji(text, replace='')",
    "        elif emoji_mode == 'text':",
    "            text = emoji.demojize(text, delimiters=(' ', ' '))",
    "",
    "    if contractions is not None:",
    "        text = contractions.fix(text)",
    "",
    "    text = re.sub(URL_PATTERN, ' ', text)",
    "    text = re.sub(MENTION_PATTERN, ' ', text)",
    "",
    "    if keep_hashtags:",
    "        text = re.sub(r'#(\\w+)', r'\\1', text)",
    "    else:",
    "        text = re.sub(HASHTAG_PATTERN, ' ', text)",
    "",
    "    text = re.sub(r'(.)\\1{2,}', r'\\1\\1', text)",
    "",
    "    if not keep_numbers:",
    "        text = re.sub(r'\\b\\d+(?:\\.\\d+)?\\b', ' ', text)",
    "",
    "    text = text.replace('_', ' ')",
    "    text = re.sub(r'[^a-z0-9\\s]', ' ', text)",
    "    text = re.sub(r'\\s+', ' ', text).strip()",
    "",
    "    return text",
]


with open(
    PROJECT_DIR
    / "preprocessing_library.py",
    "w",
    encoding="utf-8"
) as f:

    f.write(
        "\n".join(
            preprocessing_code
        )
    )


# =============================================================================
# 49. REUSABLE nltk_pipeline.py
# =============================================================================

nltk_code = [
    '"""NLTK preprocessing pipeline."""',
    "",
    "from nltk.tokenize import word_tokenize",
    "from nltk.corpus import stopwords, wordnet",
    "from nltk.stem import PorterStemmer, SnowballStemmer, WordNetLemmatizer",
    "from nltk import pos_tag",
    "from preprocessing_library import clean_text",
    "",
    "NEGATIONS = {'not', 'no', 'nor', 'never'}",
    "CUSTOM = {'amp', 'rt', 'via', 'tweet', 'tweets'}",
    "",
    "STOPWORDS = set(stopwords.words('english'))",
    "STOPWORDS -= NEGATIONS",
    "STOPWORDS.update(CUSTOM)",
    "",
    "porter = PorterStemmer()",
    "snowball = SnowballStemmer('english')",
    "lemmatizer = WordNetLemmatizer()",
    "",
    "",
    "def wn_pos(tag):",
    "    if tag.startswith('J'): return wordnet.ADJ",
    "    if tag.startswith('V'): return wordnet.VERB",
    "    if tag.startswith('N'): return wordnet.NOUN",
    "    if tag.startswith('R'): return wordnet.ADV",
    "    return wordnet.NOUN",
    "",
    "",
    "def preprocess(text):",
    "    text = clean_text(text)",
    "    tokens = [x for x in word_tokenize(text) if x.isalpha()]",
    "    filtered = [x for x in tokens if x not in STOPWORDS or x in NEGATIONS]",
    "    tagged = pos_tag(filtered) if filtered else []",
    "    lemmas = [lemmatizer.lemmatize(x, wn_pos(tag)) for x, tag in tagged]",
    "",
    "    return {",
    "        'tokens': tokens,",
    "        'filtered': filtered,",
    "        'porter': [porter.stem(x) for x in filtered],",
    "        'snowball': [snowball.stem(x) for x in filtered],",
    "        'lemmas': lemmas,",
    "        'pos': tagged",
    "    }",
]


with open(
    PROJECT_DIR
    / "nltk_pipeline.py",
    "w",
    encoding="utf-8"
) as f:

    f.write(
        "\n".join(
            nltk_code
        )
    )


# =============================================================================
# 50. REUSABLE spacy_pipeline.py
# =============================================================================

spacy_code = [
    '"""spaCy preprocessing pipeline."""',
    "",
    "import spacy",
    "from preprocessing_library import clean_text",
    "",
    "nlp = spacy.load('en_core_web_sm')",
    "",
    "NEGATIONS = {'not', 'no', 'nor', 'never'}",
    "",
    "",
    "def preprocess(text):",
    "",
    "    cleaned = clean_text(text)",
    "    doc = nlp(cleaned)",
    "",
    "    tokens = []",
    "    lemmas = []",
    "    pos_tags = []",
    "    entities = []",
    "",
    "    for token in doc:",
    "        if token.is_space or token.is_punct or not token.is_alpha:",
    "            continue",
    "",
    "        tokens.append(token.text.lower())",
    "        pos_tags.append((token.text, token.pos_))",
    "",
    "        if not token.is_stop or token.text.lower() in NEGATIONS:",
    "            lemmas.append(token.lemma_.lower())",
    "",
    "    for ent in doc.ents:",
    "        entities.append({'text': ent.text, 'label': ent.label_})",
    "",
    "    return {",
    "        'tokens': tokens,",
    "        'lemmas': lemmas,",
    "        'pos': pos_tags,",
    "        'entities': entities",
    "    }",
]


with open(
    PROJECT_DIR
    / "spacy_pipeline.py",
    "w",
    encoding="utf-8"
) as f:

    f.write(
        "\n".join(
            spacy_code
        )
    )


# =============================================================================
# 51. benchmark.py
# =============================================================================

benchmark_code = [
    '"""Standalone NLTK vs spaCy benchmark."""',
    "",
    "import time",
    "import pandas as pd",
    "from nltk_pipeline import preprocess as nltk_preprocess",
    "from spacy_pipeline import preprocess as spacy_preprocess",
    "",
    "",
    "def benchmark(texts, processor):",
    "    start = time.perf_counter()",
    "    for text in texts:",
    "        processor(text)",
    "    elapsed = time.perf_counter() - start",
    "    return elapsed, len(texts) / elapsed",
    "",
    "",
    "if __name__ == '__main__':",
    "    df = pd.read_csv('Tweets.csv')",
    "    texts = df['text'].dropna().astype(str).head(10000).tolist()",
    "",
    "    rows = []",
    "",
    "    for size in [1000, 5000, 10000]:",
    "        sample = texts[:size]",
    "",
    "        nt, ns = benchmark(sample, nltk_preprocess)",
    "        st, ss = benchmark(sample, spacy_preprocess)",
    "",
    "        rows.append({'Library': 'NLTK', 'Size': len(sample), 'Seconds': nt, 'Tweets_Per_Second': ns})",
    "        rows.append({'Library': 'spaCy', 'Size': len(sample), 'Seconds': st, 'Tweets_Per_Second': ss})",
    "",
    "    result = pd.DataFrame(rows)",
    "    result.to_csv('standalone_benchmark_results.csv', index=False)",
    "    print(result)",
]


with open(
    PROJECT_DIR
    / "benchmark.py",
    "w",
    encoding="utf-8"
) as f:

    f.write(
        "\n".join(
            benchmark_code
        )
    )


# =============================================================================
# 52. REQUIREMENTS
# =============================================================================

requirements = [
    "nltk",
    "spacy",
    "pandas",
    "numpy",
    "matplotlib",
    "seaborn",
    "emoji",
    "contractions",
    "wordcloud",
    "matplotlib-venn",
]


with open(
    PROJECT_DIR
    / "requirements.txt",
    "w",
    encoding="utf-8"
) as f:

    f.write(
        "\n".join(
            requirements
        )
    )


# =============================================================================
# 53. README
# =============================================================================

readme_lines = [

    "# Building and Benchmarking Text Preprocessing Pipelines using NLTK and spaCy",

    "",

    "## Overview",

    "",

    "This project develops and benchmarks text preprocessing pipelines using NLTK and spaCy on noisy Twitter text.",

    "",

    "The implementation uses a 10,000 tweet subset of the Twitter US Airline Sentiment dataset.",

    "",

    "## Features",

    "",

    "- Lowercase conversion",
    "- URL removal",
    "- Mention removal",
    "- Hashtag processing",
    "- Number removal",
    "- Punctuation removal",
    "- Emoji removal and emoji-to-text conversion",
    "- Contraction expansion",
    "- Repeated-character normalization",
    "- Stopword removal",
    "- Negation preservation",
    "- NLTK tokenization",
    "- Porter stemming",
    "- Snowball stemming",
    "- WordNet lemmatization",
    "- NLTK POS tagging",
    "- spaCy tokenization",
    "- spaCy lemmatization",
    "- spaCy POS tagging",
    "- Named Entity Recognition",
    "- Dependency parsing",
    "- Custom spaCy EntityRuler",
    "- Performance benchmarking",
    "- Memory benchmarking",
    "- Vocabulary analysis",

    "",

    "## Main Source Files",

    "",

    "- `preprocessing_library.py`",
    "- `nltk_pipeline.py`",
    "- `spacy_pipeline.py`",
    "- `benchmark.py`",

    "",

    "## Benchmarking",

    "",

    "The pipelines are benchmarked using 1,000, 5,000 and 10,000 tweets.",

    "",

    "Metrics include:",

    "",

    "- Processing time",
    "- Tweets per second",
    "- Peak memory",
    "- Average tokens",
    "- Vocabulary size",

    "",

    "## Quality Evaluation",

    "",

    "`quality_review_100_samples.csv` provides 100 examples for manual evaluation of lemmatization, POS tagging and NER.",

    "",

    "The original dataset does not contain gold-standard lemma, POS or NER labels, so the project does not fabricate accuracy percentages.",

    "",

    "## Installation",

    "",

    "Install dependencies using:",

    "",

    "`pip install -r requirements.txt`",

    "",

    "Then install the spaCy English model:",

    "",

    "`python -m spacy download en_core_web_sm`",

    "",

    "## Submission",

    "",

    "The submission ZIP intentionally excludes the original dataset, processed 10,000-row datasets, virtual environments, caches, installed language models and redundant generated files.",

    "",

    "Only source code, reports, benchmark summaries, the 100-sample quality file and representative result visualizations are included.",

]


with open(
    PROJECT_DIR
    / "README.md",
    "w",
    encoding="utf-8"
) as f:

    f.write(
        "\n".join(
            readme_lines
        )
    )


# =============================================================================
# 54. BENCHMARK REPORT
# =============================================================================

if nltk_time < spacy_time:

    faster = "NLTK"

elif spacy_time < nltk_time:

    faster = "spaCy"

else:

    faster = "Approximately equal"


report_lines = [

    "NLTK VS spaCy BENCHMARKING REPORT",

    "=" * 80,

    "",

    "1. METHODOLOGY",

    "",

    "A maximum of 10,000 Twitter US Airline Sentiment tweets were used.",

    "",

    "The same cleaning strategy was applied before NLTK and spaCy linguistic processing.",

    "",

    "2. TEXT CLEANING",

    "",

    "The cleaning stage removes URLs, mentions, numbers, punctuation and extra whitespace while processing hashtags, emojis, contractions and repeated characters.",

    "",

    "Negation terms including not, no, nor and never are preserved.",

    "",

    "3. NLTK PIPELINE",

    "",

    "NLTK uses word tokenization, custom stopword removal, Porter stemming, Snowball stemming, WordNet lemmatization and POS tagging.",

    "",

    "4. spaCy PIPELINE",

    "",

    "spaCy uses its integrated tokenizer, lemmatizer, POS tagger, dependency parser and Named Entity Recognition pipeline.",

    "",

    "A custom EntityRuler adds airline-specific entities.",

    "",

    "5. PERFORMANCE",

    "",

    "Largest benchmark: "
    + str(
        int(
            largest_size
        )
    )
    + " tweets",

    "",

    "NLTK time: "
    + format(
        nltk_time,
        ".3f"
    )
    + " seconds",

    "spaCy time: "
    + format(
        spacy_time,
        ".3f"
    )
    + " seconds",

    "",

    "Faster pipeline on this run: "
    + faster,

    "",

    "Performance depends on CPU, RAM, package versions and runtime conditions.",

    "",

    "6. VOCABULARY",

    "",

    "Original vocabulary: "
    + str(
        len(
            original_vocabulary
        )
    ),

    "NLTK vocabulary: "
    + str(
        len(
            nltk_vocabulary
        )
    ),

    "spaCy vocabulary: "
    + str(
        len(
            spacy_vocabulary
        )
    ),

    "",

    "Preprocessing reduces vocabulary by removing noise and normalizing word forms.",

    "",

    "7. LEMMATIZATION",

    "",

    "NLTK WordNet lemmatization uses POS information in this implementation.",

    "",

    "spaCy performs lemmatization as part of its integrated linguistic pipeline.",

    "",

    "8. NER",

    "",

    "spaCy provides built-in NER. Social-media NER remains challenging because tweets frequently contain slang, abbreviations, spelling errors, usernames and inconsistent capitalization.",

    "",

    "9. STEMMING VS LEMMATIZATION",

    "",

    "Stemming is useful when aggressive vocabulary reduction is acceptable. Lemmatization is generally preferable when linguistically meaningful base forms are required.",

    "",

    "10. NLTK USE CASES",

    "",

    "NLTK is particularly useful for teaching, research, algorithm experimentation and highly customized preprocessing.",

    "",

    "11. spaCy USE CASES",

    "",

    "spaCy is particularly useful for integrated POS tagging, lemmatization, NER, dependency parsing, batch processing and production NLP.",

    "",

    "12. HYBRID APPROACH",

    "",

    "A hybrid approach can combine NLTK lexical resources and custom preprocessing with spaCy linguistic analysis.",

    "",

    "13. CONCLUSION",

    "",

    "Neither library is universally better. NLTK provides fine-grained algorithmic control, while spaCy provides a more integrated production-oriented NLP pipeline.",

]


with open(
    PROJECT_DIR
    / "benchmark_report.txt",
    "w",
    encoding="utf-8"
) as f:

    f.write(
        "\n".join(
            report_lines
        )
    )


# =============================================================================
# 55. SAVE A COPY OF THIS NOTE AS main.py
# =============================================================================

# If running from Jupyter, Python cannot automatically recover the entire
# notebook cell source reliably.
#
# We therefore create a small main.py launcher explaining the modules.

main_lines = [

    '"""',
    "Main entry information for the NLTK vs spaCy preprocessing project.",
    "",
    "The complete experiment was executed in Jupyter.",
    "Reusable functionality is provided in:",
    "",
    "- preprocessing_library.py",
    "- nltk_pipeline.py",
    "- spacy_pipeline.py",
    "- benchmark.py",
    '"""',

    "",

    "print('Text Preprocessing Pipelines using NLTK and spaCy')",

    "print('Reusable modules:')",

    "print('1. preprocessing_library.py')",

    "print('2. nltk_pipeline.py')",

    "print('3. spacy_pipeline.py')",

    "print('4. benchmark.py')",

]


with open(
    PROJECT_DIR
    / "main.py",
    "w",
    encoding="utf-8"
) as f:

    f.write(
        "\n".join(
            main_lines
        )
    )


# =============================================================================
# 56. CREATE SLIM SUBMISSION DIRECTORY
# =============================================================================

print(
    "\n" + "=" * 100
)

print(
    "CREATING SLIM SUBMISSION"
)

print(
    "=" * 100
)


SLIM_DIR = (
    PROJECT_DIR
    / "SLIM_SUBMISSION"
)


if SLIM_DIR.exists():

    shutil.rmtree(
        SLIM_DIR
    )


SLIM_DIR.mkdir(
    parents=True,
    exist_ok=True
)


SCREENSHOT_DIR = (
    SLIM_DIR
    / "screenshots"
)


SCREENSHOT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# =============================================================================
# 57. ONLY ESSENTIAL SOURCE/REPORT FILES
# =============================================================================

ESSENTIAL_FILES = [

    "main.py",

    "preprocessing_library.py",

    "nltk_pipeline.py",

    "spacy_pipeline.py",

    "benchmark.py",

    "README.md",

    "requirements.txt",

    "benchmark_report.txt",

    "benchmark_results.csv",

    "preprocessing_comparison.csv",

    "quality_review_100_samples.csv",

    "stemming_vs_lemmatization.csv",

    "vocabulary_statistics.csv",

    "entity_frequency.csv",

]


for filename in ESSENTIAL_FILES:

    source = (
        PROJECT_DIR
        / filename
    )


    if source.exists():

        shutil.copy2(
            source,
            SLIM_DIR
            / filename
        )


# =============================================================================
# 58. REPRESENTATIVE SCREENSHOTS ONLY
# =============================================================================

ESSENTIAL_SCREENSHOTS = [

    "tweet_length_distribution.png",

    "processing_time_comparison.png",

    "token_count_distribution.png",

    "nltk_pos_distribution.png",

    "spacy_pos_distribution.png",

    "named_entity_frequency.png",

    "stopword_removal_impact.png",

    "stemming_lemmatization_comparison.png",

    "vocabulary_overlap.png",

    "dependency_parse_example.png",

]


for filename in ESSENTIAL_SCREENSHOTS:

    source = (
        PROJECT_DIR
        / filename
    )


    if source.exists():

        shutil.copy2(
            source,
            SCREENSHOT_DIR
            / filename
        )


# =============================================================================
# 59. EXPLICITLY VERIFY LARGE FILES ARE NOT IN SUBMISSION
# =============================================================================

FORBIDDEN_NAMES = {

    "Tweets.csv",

    "database.sqlite",

    "nltk_processed_tweets.csv",

    "spacy_processed_tweets.csv",

    "final_preprocessing_results.csv",

    "extracted_entities.json",

    "wordcloud_before.png",

    "wordcloud_after.png",

}


for path in (
    SLIM_DIR.rglob(
        "*"
    )
):

    if path.name in FORBIDDEN_NAMES:

        if path.is_file():

            path.unlink()


# =============================================================================
# 60. REMOVE CACHE/ENV FOLDERS IF THEY SOMEHOW EXIST
# =============================================================================

FORBIDDEN_DIR_NAMES = {

    "__pycache__",

    ".cache",

    ".pytest_cache",

    ".ipynb_checkpoints",

    "venv",

    ".venv",

    "env",

    "node_modules",

    "build",

    "dist",

}


for path in list(
    SLIM_DIR.rglob(
        "*"
    )
):

    if (
        path.is_dir()
        and
        path.name
        in FORBIDDEN_DIR_NAMES
    ):

        shutil.rmtree(
            path,
            ignore_errors=True
        )


# =============================================================================
# 61. SLIM MANIFEST
# =============================================================================

manifest = []


for path in sorted(
    SLIM_DIR.rglob(
        "*"
    )
):

    if path.is_file():

        size_kb = (
            path.stat().st_size
            / 1024
        )


        manifest.append({

            "File":
                str(
                    path.relative_to(
                        SLIM_DIR
                    )
                ),

            "Size_KB":
                round(
                    size_kb,
                    2
                )

        })


manifest_df = pd.DataFrame(
    manifest
)


manifest_df.to_csv(
    SLIM_DIR
    / "submission_manifest.csv",
    index=False
)


print(
    "\nFiles going into submission:"
)

print(
    manifest_df.to_string(
        index=False
    )
)


# =============================================================================
# 62. CHECK SLIM DIRECTORY SIZE
# =============================================================================

total_bytes = sum(

    path.stat().st_size

    for path in (
        SLIM_DIR.rglob(
            "*"
        )
    )

    if path.is_file()

)


total_mb = (
    total_bytes
    / 1024
    / 1024
)


print(
    "\nSLIM submission folder size:",
    round(
        total_mb,
        2
    ),
    "MB"
)


# =============================================================================
# 63. CREATE SLIM ZIP
# =============================================================================

SLIM_ZIP = (
    PROJECT_DIR.parent
    / "Text_Preprocessing_NLTK_spaCy_SLIM.zip"
)


if SLIM_ZIP.exists():

    SLIM_ZIP.unlink()


with zipfile.ZipFile(
    SLIM_ZIP,
    "w",
    compression=zipfile.ZIP_DEFLATED,
    compresslevel=9
) as zf:

    for file_path in (
        SLIM_DIR.rglob(
            "*"
        )
    ):

        if not file_path.is_file():

            continue


        relative_path = (
            file_path.relative_to(
                SLIM_DIR
            )
        )


        zf.write(
            file_path,
            arcname=relative_path
        )


zip_size_mb = (
    SLIM_ZIP.stat().st_size
    / 1024
    / 1024
)


# =============================================================================
# 64. ZIP CONTENT VERIFICATION
# =============================================================================

print(
    "\nZIP CONTENTS:"
)


with zipfile.ZipFile(
    SLIM_ZIP,
    "r"
) as zf:

    zip_names = (
        zf.namelist()
    )


    for name in zip_names:

        print(
            " [OK]",
            name
        )


# =============================================================================
# 65. ENSURE DATASET NOT PRESENT
# =============================================================================

bad_files = []


for name in zip_names:

    base_name = (
        Path(
            name
        ).name
    )


    if (
        base_name
        in FORBIDDEN_NAMES
    ):

        bad_files.append(
            name
        )


if bad_files:

    print(
        "\nWARNING: unwanted files found:"
    )

    print(
        bad_files
    )

else:

    print(
        "\n[PASS] No raw datasets or large processed datasets are inside the ZIP."
    )


# =============================================================================
# 66. FINAL RESULTS
# =============================================================================

print(
    "\n" + "=" * 100
)

print(
    "PROJECT COMPLETED SUCCESSFULLY"
)

print(
    "=" * 100
)


print(
    "\nFULL RESULTS:"
)

print(
    PROJECT_DIR
)


print(
    "\nSLIM SUBMISSION FOLDER:"
)

print(
    SLIM_DIR
)


print(
    "\nUPLOAD THIS ZIP:"
)

print(
    SLIM_ZIP
)


print(
    "\nZIP SIZE:",
    round(
        zip_size_mb,
        2
    ),
    "MB"
)


if zip_size_mb < 10:

    print(
        "\n[EXCELLENT] ZIP is very small and suitable for submission."
    )

elif zip_size_mb < 25:

    print(
        "\n[PASS] ZIP is reasonably small."
    )

elif zip_size_mb < 50:

    print(
        "\n[PASS] ZIP is under 50 MB, although further reduction is possible."
    )

else:

    print(
        "\n[WARNING] ZIP is still above 50 MB."
    )


print(
    "\nThe following are intentionally NOT included in the ZIP:"
)

print(
    "- Tweets.csv"
)

print(
    "- database.sqlite"
)

print(
    "- nltk_processed_tweets.csv"
)

print(
    "- spacy_processed_tweets.csv"
)

print(
    "- final_preprocessing_results.csv"
)

print(
    "- extracted_entities.json"
)

print(
    "- WordCloud images"
)

print(
    "- virtual environments"
)

print(
    "- spaCy model files"
)

print(
    "- cache folders"
)

print(
    "- Jupyter output/cache files"
)


print(
    "\nThese full results still remain available in:"
)

print(
    PROJECT_DIR
)


print(
    "\n" + "=" * 100
)

print(
    "UPLOAD Text_Preprocessing_NLTK_spaCy_SLIM.zip"
)

print(
    "=" * 100
)
```
